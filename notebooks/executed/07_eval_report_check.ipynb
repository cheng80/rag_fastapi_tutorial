## 튜토리얼 앱 연결

이제 작은 실험으로 튜토리얼 앱의 어느 파일과 이어지는지 확인합니다. 코드가 길어 보여도 볼 것은 하나입니다. 출력이 예상과 다르면 바로 앞 셀부터 다시 확인하세요.

### 1. eval 파일 목록

이 셀에서는 `eval 파일 목록` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
from pathlib import Path
ROOT = Path.cwd()
EVAL_ROOT = ROOT / 'project_template' / 'data' / 'eval'
files = sorted(p.name for p in EVAL_ROOT.glob('*.jsonl'))
print('count:', len(files))
print(files[:30])
assert 'tourism_20_questions.jsonl' in files
```

### 2. 20문항 샘플 읽기

이 셀에서는 `20문항 샘플 읽기` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
import json
path = EVAL_ROOT / 'tourism_20_questions.jsonl'
rows = [json.loads(line) for line in path.read_text(encoding='utf-8').splitlines() if line.strip()]
print('rows:', len(rows))
print(rows[0])
assert len(rows) >= 20
```

### 3. 질문 필드 분포 관찰

이 셀에서는 `질문 필드 분포 관찰` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
keys = sorted({key for row in rows for key in row})
print(keys)
assert keys
```

### 4. 이벤트 로그 서비스 위치

이 셀에서는 `이벤트 로그 서비스 위치` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
service = ROOT / 'project_template' / 'app' / 'services' / 'tourism_query_event_logger.py'
text = service.read_text(encoding='utf-8')
print(text[:500])
assert 'event' in text.lower() or 'query' in text.lower()
```

### 5. 테스트 파일에서 검증 의도 찾기

이 셀에서는 `테스트 파일에서 검증 의도 찾기` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
tests = sorted((ROOT / 'project_template' / 'tests').glob('test_*.py'))
for test in tests[:20]:
    print(test.name)
assert tests
```

### 6. 운영 리포트 체크리스트

이 셀에서는 `운영 리포트 체크리스트` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.
