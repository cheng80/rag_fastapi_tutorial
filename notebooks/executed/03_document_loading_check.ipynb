## 원본형 앱 연결

아래 셀부터는 작은 실험을 통해 원본형 앱의 어느 파일과 연결되는지 확인합니다. 코드가 길어 보여도 목적은 하나입니다. 초급자는 출력된 값이 기대와 다른 순간 바로 이전 셀로 돌아가면 됩니다.

### 1. 데이터 폴더 지도

이 셀은 `데이터 폴더 지도`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from pathlib import Path
PROJECT_ROOT = Path.cwd()
DATA_ROOT = PROJECT_ROOT / 'project_template' / 'data'
for child in sorted(DATA_ROOT.iterdir()):
    print(child.name, 'dir' if child.is_dir() else 'file')
assert (DATA_ROOT / 'raw').exists()
```

### 2. raw markdown 읽기

이 셀은 `raw markdown 읽기`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
raw_file = DATA_ROOT / 'raw' / 'example_faq.md'
text = raw_file.read_text(encoding='utf-8')
print(text[:500])
assert len(text) > 100
```

### 3. processed JSON 목록

이 셀은 `processed JSON 목록`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
processed = DATA_ROOT / 'processed'
json_files = sorted(p.name for p in processed.glob('*.json'))
print(json_files[:20])
assert 'tour_area_codes.json' in json_files
```

### 4. 지역 코드 샘플 확인

이 셀은 `지역 코드 샘플 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import json
area_codes = json.loads((processed / 'tour_area_codes.json').read_text(encoding='utf-8'))
print(type(area_codes).__name__)
print(str(area_codes)[:500])
assert area_codes
```

### 5. eval JSONL 샘플 확인

이 셀은 `eval JSONL 샘플 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
eval_file = DATA_ROOT / 'eval' / 'tourism_20_questions.jsonl'
lines = eval_file.read_text(encoding='utf-8').splitlines()
print('lines:', len(lines))
print(lines[0][:500])
assert len(lines) >= 20
```

### 6. 문서 로딩 체크리스트

이 셀은 `문서 로딩 체크리스트`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
checklist = ['raw text exists', 'processed codes exist', 'eval questions exist']
for item in checklist:
    print('[문서 로딩]', item)
assert len(checklist) == 3
```
