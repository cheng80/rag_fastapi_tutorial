## 원본형 앱 연결

아래 셀부터는 작은 실험을 통해 원본형 앱의 어느 파일과 연결되는지 확인합니다. 코드가 길어 보여도 목적은 하나입니다. 초급자는 출력된 값이 기대와 다른 순간 바로 이전 셀로 돌아가면 됩니다.

### 1. import 경로 준비

이 셀은 `import 경로 준비`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd()
TEMPLATE_ROOT = PROJECT_ROOT / 'project_template'
sys.path.insert(0, str(TEMPLATE_ROOT))
print('import root:', TEMPLATE_ROOT)
```

### 2. 앱 객체 import

이 셀은 `앱 객체 import`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from app.main import app
print(type(app).__name__)
assert app.title
```

### 3. 등록 route 살펴보기

이 셀은 `등록 route 살펴보기`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
routes = sorted(getattr(route, 'path', '') for route in app.routes)
for route in routes:
    print(route)
assert '/health' in routes
```

### 4. health 요청

이 셀은 `health 요청`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from fastapi.testclient import TestClient
client = TestClient(app)
response = client.get('/health')
print(response.status_code, response.json())
assert response.status_code == 200
```

### 5. OpenAPI 표면 확인

이 셀은 `OpenAPI 표면 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
schema = client.get('/openapi.json').json()
paths = sorted(schema.get('paths', {}))
print(paths[:20])
assert '/health' in paths
```

### 6. 정적 Web UI 연결 확인

이 셀은 `정적 Web UI 연결 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
response = client.get('/tourism-ui')
print(response.status_code, response.text[:120])
assert response.status_code == 200
html = response.text
assert '<html' in html.lower() or '<!doctype html' in html.lower()
```
