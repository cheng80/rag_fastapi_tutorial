## 튜토리얼 앱 연결

이제 작은 실험으로 튜토리얼 앱의 어느 파일과 이어지는지 확인합니다. 코드가 길어 보여도 볼 것은 하나입니다. 출력이 예상과 다르면 바로 앞 셀부터 다시 확인하세요.

### 1. import 경로 준비

이 셀에서는 `import 경로 준비` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd()
TEMPLATE_ROOT = PROJECT_ROOT / 'project_template'
sys.path.insert(0, str(TEMPLATE_ROOT))
print('import root:', TEMPLATE_ROOT)
```

### 2. 앱 객체 import

이 셀에서는 `앱 객체 import` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
from app.main import app
print(type(app).__name__)
assert app.title
```

### 3. 등록 route 살펴보기

이 셀에서는 `등록 route 살펴보기` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
routes = sorted(getattr(route, 'path', '') for route in app.routes)
for route in routes:
    print(route)
assert '/health' in routes
```

### 4. health 요청

이 셀에서는 `health 요청` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
from fastapi.testclient import TestClient
client = TestClient(app)
response = client.get('/health')
print(response.status_code, response.json())
assert response.status_code == 200
```

### 5. OpenAPI 응답 확인

이 셀에서는 `OpenAPI 응답 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
schema = client.get('/openapi.json').json()
paths = sorted(schema.get('paths', {}))
print(paths[:20])
assert '/health' in paths
```

### 6. 정적 Web UI 연결 확인

이 셀에서는 `정적 Web UI 연결 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
response = client.get('/tourism-ui')
print(response.status_code, response.text[:120])
assert response.status_code == 200
html = response.text
assert '<html' in html.lower() or '<!doctype html' in html.lower()
```
