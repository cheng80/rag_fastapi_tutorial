## 튜토리얼 앱 연결

### 1. 프로젝트 루트 찾기

이 셀에서는 `프로젝트 루트 찾기` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
from pathlib import Path
PROJECT_ROOT = Path.cwd()
TEMPLATE_ROOT = PROJECT_ROOT / 'project_template'
print('project root:', PROJECT_ROOT.name)
print('template exists:', TEMPLATE_ROOT.exists())
assert TEMPLATE_ROOT.exists()
```

### 2. Python 버전 확인

이 셀에서는 `Python 버전 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
import sys
print(sys.version)
assert sys.version_info >= (3, 10)
```

### 3. 핵심 파일 확인

이 셀에서는 `핵심 파일 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
required = ['app/main.py', 'app/core/config.py', 'frontend/web/index.html', 'requirements.txt']
for name in required:
    path = TEMPLATE_ROOT / name
    print(name, path.exists())
    assert path.exists(), name
```

### 4. 환경 파일 정책 확인

이 셀에서는 `환경 파일 정책 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
env_candidates = [TEMPLATE_ROOT / '.env.example', TEMPLATE_ROOT / '.env.sample']
env_file = TEMPLATE_ROOT / '.env'
print('env examples:', {path.name: path.exists() for path in env_candidates})
print('.env local exists:', env_file.exists())
print('튜토리얼 템플릿은 개인 .env 값을 복사하지 않습니다. 필요한 값은 로컬 .env에만 둡니다.')
```

### 5. 의존성 import smoke

이 셀에서는 `의존성 import smoke` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
import importlib.util
for module_name in ['fastapi', 'pydantic', 'pytest']:
    print(module_name, importlib.util.find_spec(module_name) is not None)
```
