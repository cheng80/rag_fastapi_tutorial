## 원본형 앱 연결

아래 셀부터는 작은 실험을 통해 원본형 앱의 어느 파일과 연결되는지 확인합니다. 코드가 길어 보여도 목적은 하나입니다. 초급자는 출력된 값이 기대와 다른 순간 바로 이전 셀로 돌아가면 됩니다.

### 1. 프로젝트 루트 찾기

이 셀은 `프로젝트 루트 찾기`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from pathlib import Path
PROJECT_ROOT = Path.cwd()
TEMPLATE_ROOT = PROJECT_ROOT / 'project_template'
print('project root:', PROJECT_ROOT.name)
print('template exists:', TEMPLATE_ROOT.exists())
assert TEMPLATE_ROOT.exists()
```

### 2. Python 버전 확인

이 셀은 `Python 버전 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import sys
print(sys.version)
assert sys.version_info >= (3, 10)
```

### 3. 핵심 파일 확인

이 셀은 `핵심 파일 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
required = ['app/main.py', 'app/core/config.py', 'frontend/web/index.html', 'requirements.txt']
for name in required:
    path = TEMPLATE_ROOT / name
    print(name, path.exists())
    assert path.exists(), name
```

### 4. 환경 파일 정책 확인

이 셀은 `환경 파일 정책 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
env_candidates = [TEMPLATE_ROOT / '.env.example', TEMPLATE_ROOT / '.env.sample']
env_file = TEMPLATE_ROOT / '.env'
print('env examples:', {path.name: path.exists() for path in env_candidates})
print('.env local exists:', env_file.exists())
print('원본형 템플릿은 개인 .env 값을 복사하지 않습니다. 필요한 값은 로컬 .env에만 둡니다.')
```

### 5. 의존성 import smoke

이 셀은 `의존성 import smoke`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import importlib.util
for module_name in ['fastapi', 'pydantic', 'pytest']:
    print(module_name, importlib.util.find_spec(module_name) is not None)
```

### 6. 학습자가 기록할 상태표

이 셀은 `학습자가 기록할 상태표`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
status = {
    'path_ready': TEMPLATE_ROOT.exists(),
    'env_policy_ready': True,
    'can_start_fastapi_chapter': True,
}
print(status)
assert status['can_start_fastapi_chapter']
```
