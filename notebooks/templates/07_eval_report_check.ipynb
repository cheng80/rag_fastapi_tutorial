## 원본형 앱 연결

아래 셀부터는 작은 실험을 통해 원본형 앱의 어느 파일과 연결되는지 확인합니다. 코드가 길어 보여도 목적은 하나입니다. 초급자는 출력된 값이 기대와 다른 순간 바로 이전 셀로 돌아가면 됩니다.

### 1. eval 파일 목록

이 셀은 `eval 파일 목록`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from pathlib import Path
ROOT = Path.cwd()
EVAL_ROOT = ROOT / 'project_template' / 'data' / 'eval'
files = sorted(p.name for p in EVAL_ROOT.glob('*.jsonl'))
print('count:', len(files))
print(files[:30])
assert 'tourism_20_questions.jsonl' in files
```

### 2. 20문항 샘플 읽기

이 셀은 `20문항 샘플 읽기`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import json
path = EVAL_ROOT / 'tourism_20_questions.jsonl'
rows = [json.loads(line) for line in path.read_text(encoding='utf-8').splitlines() if line.strip()]
print('rows:', len(rows))
print(rows[0])
assert len(rows) >= 20
```

### 3. 질문 필드 분포 관찰

이 셀은 `질문 필드 분포 관찰`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
keys = sorted({key for row in rows for key in row})
print(keys)
assert keys
```

### 4. 이벤트 로그 서비스 위치

이 셀은 `이벤트 로그 서비스 위치`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
service = ROOT / 'project_template' / 'app' / 'services' / 'tourism_query_event_logger.py'
text = service.read_text(encoding='utf-8')
print(text[:500])
assert 'event' in text.lower() or 'query' in text.lower()
```

### 5. 테스트 파일에서 검증 의도 찾기

이 셀은 `테스트 파일에서 검증 의도 찾기`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
tests = sorted((ROOT / 'project_template' / 'tests').glob('test_*.py'))
for test in tests[:20]:
    print(test.name)
assert tests
```

### 6. 운영 리포트 체크리스트

이 셀은 `운영 리포트 체크리스트`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
report_checks = ['dataset count', 'status code', 'card evidence', 'message_included_count', 'browser surface']
for check in report_checks:
    print('[eval]', check)
assert 'browser surface' in report_checks
```
