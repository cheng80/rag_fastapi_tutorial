## 튜토리얼 앱 연결

이제 작은 실험으로 튜토리얼 앱의 어느 파일과 이어지는지 확인합니다. 코드가 길어 보여도 볼 것은 하나입니다. 출력이 예상과 다르면 바로 앞 셀부터 다시 확인하세요.

### 1. 데이터 폴더 지도

이 셀에서는 `데이터 폴더 지도` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
from pathlib import Path
PROJECT_ROOT = Path.cwd()
DATA_ROOT = PROJECT_ROOT / 'project_template' / 'data'
for child in sorted(DATA_ROOT.iterdir()):
    print(child.name, 'dir' if child.is_dir() else 'file')
assert (DATA_ROOT / 'raw').exists()
```

### 2. raw markdown 읽기

이 셀에서는 `raw markdown 읽기` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
raw_file = DATA_ROOT / 'raw' / 'example_faq.md'
text = raw_file.read_text(encoding='utf-8')
print(text[:500])
assert len(text) > 100
```

### 3. processed JSON 목록

이 셀에서는 `processed JSON 목록` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
processed = DATA_ROOT / 'processed'
json_files = sorted(p.name for p in processed.glob('*.json'))
print(json_files[:20])
assert 'tour_area_codes.json' in json_files
```

### 4. 지역 코드 샘플 확인

이 셀에서는 `지역 코드 샘플 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
import json
area_codes = json.loads((processed / 'tour_area_codes.json').read_text(encoding='utf-8'))
print(type(area_codes).__name__)
print(str(area_codes)[:500])
assert area_codes
```

### 5. eval JSONL 샘플 확인

이 셀에서는 `eval JSONL 샘플 확인` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
eval_file = DATA_ROOT / 'eval' / 'tourism_20_questions.jsonl'
lines = eval_file.read_text(encoding='utf-8').splitlines()
print('lines:', len(lines))
print(lines[0][:500])
assert len(lines) >= 20
```

### 6. 문서 로딩 체크리스트

이 셀에서는 `문서 로딩 체크리스트` 항목을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 쓰세요.

```python
checklist = ['raw text exists', 'processed codes exist', 'eval questions exist']
for item in checklist:
    print('[문서 로딩]', item)
assert len(checklist) == 3
```
