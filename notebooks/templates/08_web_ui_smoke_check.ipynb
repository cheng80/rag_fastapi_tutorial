## 원본형 앱 연결

아래 셀부터는 작은 실험을 통해 원본형 앱의 어느 파일과 연결되는지 확인합니다. 코드가 길어 보여도 목적은 하나입니다. 초급자는 출력된 값이 기대와 다른 순간 바로 이전 셀로 돌아가면 됩니다.

### 1. 프론트엔드 파일 지도

이 셀은 `프론트엔드 파일 지도`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
from pathlib import Path
ROOT = Path.cwd()
FRONTEND = ROOT / 'project_template' / 'frontend'
for path in sorted(FRONTEND.rglob('*')):
    if path.is_file():
        print(path.relative_to(FRONTEND))
assert (FRONTEND / 'web' / 'index.html').exists()
```

### 2. HTML 텍스트 확인

이 셀은 `HTML 텍스트 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
html = (FRONTEND / 'web' / 'index.html').read_text(encoding='utf-8')
print(html[:700])
assert '튜토리얼 결과' not in html
assert '<script' in html.lower()
```

### 3. JS 자산 확인

이 셀은 `JS 자산 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
js_files = sorted(FRONTEND.rglob('*.js'))
for path in js_files:
    text = path.read_text(encoding='utf-8')
    print(path.name, len(text))
assert js_files
```

### 4. 정적 페이지 HTTP smoke

이 셀은 `정적 페이지 HTTP smoke`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
import sys
from fastapi.testclient import TestClient
TEMPLATE_ROOT = ROOT / 'project_template'
sys.path.insert(0, str(TEMPLATE_ROOT))
from app.main import app
client = TestClient(app)
response = client.get('/tourism-ui')
print(response.status_code, response.text[:200])
assert response.status_code == 200
```

### 5. 브라우저 QA 증거 파일 확인

이 셀은 `브라우저 QA 증거 파일 확인`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
evidence_root = ROOT / '.omo' / 'ulw-loop'
files = sorted(evidence_root.rglob('C003-browser-ui*')) if evidence_root.exists() else []
for path in files:
    print(path.relative_to(ROOT))
assert files
```

### 6. 릴리스 표면 체크리스트

이 셀은 `릴리스 표면 체크리스트`을 확인합니다. 출력은 다음 장으로 넘어가기 전 체크리스트로 사용합니다.

```python
surface_checks = ['no tutorial wording', 'chat input visible', 'tourism cards render', 'evidence expandable', 'debug hidden in release']
for item in surface_checks:
    print('[web-ui]', item)
assert 'no tutorial wording' in surface_checks
```
